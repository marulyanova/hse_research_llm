```python
import json
import random

set_seed()

with open("train.json") as f:
    train_dataset = json.load(f)

with open("val.json") as f:
    val_dataset = json.load(f)

train_dataset = random.sample(train_dataset, 250)
val_dataset = random.sample(val_dataset, 50)

def calc_difficulty(dataset):
    difficulties = {}
    for item in dataset:
        if item["difficulty"] not in difficulties:
            difficulties[item["difficulty"]] = 1
        else:
            difficulties[item["difficulty"]] += 1
    return sorted(difficulties.items(), key = lambda x: x[0])

print(f"Train Data: {len(train_dataset)}, difficulties: {calc_difficulty(train_dataset)}")
print(f"Val Data: {len(val_dataset)}, difficulties: {calc_difficulty(val_dataset)}")
```

```python
grpo_train_dataset = prepare_grpo_dataset(train_dataset)

trainer = GRPOTrainer(
    model=grpo_model,
    processing_class=grpo_tokenizer,
    reward_funcs=[make_reward_function(grpo_model, grpo_tokenizer, verifier, grpo_train_dataset)],
    args=grpo_config,
    train_dataset=grpo_train_dataset,
)
```
